```python
import pandas as pd, numpy as np
from faker import Faker
import uuid, random, os
from datetime import datetime, timedelta

fake = Faker('en_IN')
np.random.seed(42); random.seed(42)

categories = ['Food','Groceries','Recharge','EMI',
              'Shopping','Transport','Utilities','Medical']
cat_wts = [0.22,0.18,0.12,0.10,0.16,0.10,0.08,0.04]
users   = [str(uuid.uuid4())[:8] for _ in range(500)]
start   = datetime(2024,1,1)

rows = []
for _ in range(10000):
    ts  = start + timedelta(days=random.randint(0,364),
                hours=random.randint(0,23),
                minutes=random.randint(0,59))
    amt = round(min(max(np.random.lognormal(6.5,1.2),10),50000),2)
    rows.append({
        'transaction_id': str(uuid.uuid4()),
        'user_id': random.choice(users),
        'timestamp': ts, 'amount': amt,
        'merchant_category': random.choices(categories,cat_wts)[0],
        'city_tier': random.choices(['Tier 1','Tier 2','Tier 3'],
                                    [0.45,0.35,0.20])[0],
        'payment_status': random.choices(['Success','Failed','Pending'],
                                         [0.88,0.09,0.03])[0],
        'device_type': random.choices(['Android','iOS','Web'],
                                      [0.72,0.20,0.08])[0],
    })
df = pd.DataFrame(rows)
os.makedirs('data/raw', exist_ok=True)
df.to_csv('data/raw/upi_transactions.csv', index=False)
print(f"Done! {len(df):,} rows saved to data/raw/upi_transactions.csv")
print(df.head(3))
```
